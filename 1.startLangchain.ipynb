## 开发流程

1. 加载环境变量
2. 定义工具
3. 定义Agent
4. 调用Agent

### 1.加载环境变量

```python
from dotenv import load_dotenv
load_dotenv()
```

### 2. 定义工具

```python
## 定义工具
from langchain.tools import tool

@tool
def getWeather(location: str) -> str:
    """根据传入的地点名称，查询该地实时天气状况."""
    return f"{location} 当前天气晴朗，最高气温 25 摄氏度."
```

### 3.创建Agent

```python
## 3.创建Agent
from langchain_deepseek import ChatDeepSeek
from langchain.agents import create_agent

llm = ChatDeepSeek(model="deepseek-chat")  # 直接替换 v4-flash



agent = create_agent(llm, tools=[getWeather])
```

### 4.发起调用

```python
response = agent.invoke({"messages": [{"role": "user", "content": "太原的天气怎么样?"}]})
for message in response['messages']:
    print(message.model_dump_json(indent=2))
    print("==================================")
```
