```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain.messages import HumanMessage
from langchain.agents import create_agent

load_dotenv()

api_key = os.getenv("DEEPSEEK_API_KEY")
base_url = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="openai",
    temperature=0.1,
    api_key=api_key,
    base_url=base_url
)

system_prompt = """
# 身份  
- 你是一名网络安全红队专家,你可以帮助用户完成渗透测试工作,并给出安全工具的命令提示  


# 指令  

- 输出信息时不能返回markdown格式的说明,只返回可用的命令与命令的注释


"""
agent = create_agent(
    model=model,
    system_prompt=system_prompt,
    tools=[]
)
if not api_key:
    raise ValueError("DEEPSEEK_API_KEY 未设置，请在 .env 文件中配置")

for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="请帮我进行渗透测试,并给出命令提示")]},
    stream_mode="messages"
):
    print(token.content, end="", flush=True)if not base_url:
    raise ValueError("DEEPSEEK_BASE_URL 未设置，请在 .env 文件中配置")

try:
    agent = create_agent(
        model=model,
        system_prompt=system_prompt,
        tools=[]
    )
except Exception as e:
    raise RuntimeError(f"创建 Agent 失败: {e}")

try:
    for token, metadata in agent.stream(
        {"messages": [HumanMessage(content="请帮我进行渗透测试,并给出命令提示")]},
        stream_mode="messages"
    ):
        print(token.content, end="", flush=True)
except Exception as e:
    print(f"\n流式输出出错: {e}")
```
